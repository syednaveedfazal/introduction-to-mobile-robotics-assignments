## 4.1 Implement model for perceiving landmarks [2.0]

Consider a robot that operates in an environment populated by **range-only** sensors such as _beacons_.
These sensors transmit a signal that allows for obtaining noisy measurements of their distance from a receiver mounted on the robot. For this type of navigation, the location of beacons is assumed to be known.

Each beacon may have different reliability or measurement characteristics: for example, a large landmark, such as a mountain, may be detected more easily, but the measured distance to it can be less precise than that to a small, well defined street sign. This variablity is captured in the observation model as per-beacon-measurment noise $\sigma_z$.

The function `landmark_observation_model` realizes the observation model for perceiving landmarks with a range sensor. It takes as input:
 
- the observed noisy range $z$
- the standard deviation of the measurement $\sigma_z$ (zero-mean Gaussian noise)
- the position of the transmitting beacon in the map $b$
- the position of the robot in the map $x$

The output of this function is the probability $p(z | x, m)$ that the measurement $z$ was obtained at position $x$.

Implement the `landmark_observation_model` function.

**Hint:** The likelihood for the first measurement should  be roughly 10 times higher!

```python
z1 = 2.5
z2 = 5
x = np.array([ 0, 0 ])
b = np.array([ 2, 2 ])
sigma_z1 = 1
sigma_z2 = 1

likelihood1 = landmark_observation_model(z1, sigma_z1, b, x)
likelihood2 = landmark_observation_model(z2, sigma_z2, b, x)

print(f"Likelihood for observation z={z1} is {likelihood1}.")
print(f"Likelihood for observation z={z2} is {likelihood2}.")
```

## 4.3 Multiple landmarks [2.0]

Assume that we install two **additional** beacons in our environment:

- $b_2 = [8, 18]$
- $b_3 = [16, 8]$

from which we obtain the following observations:

- $z_2 = 5$ with $\sigma_{z2} = 5$
- $z_3 = 9$ with $\sigma_{z3} = 3$

Complete the `joint_observation_likelihood` function to compute the `observation_likelihood` for each cell of the environment by including one ore more observations. Make sure to return the **joint** likelihood of the given observations. 

**Hint:** Each additional beacon reduces the uncertainty in the location - the amount depends on the measurment noise for the specific beacon!

```python
#Gridmap
size = 20
gridmap = np.zeros([size, size])

#Beacons
beacon1 = np.array([6,2])
beacon2 = np.array([8,18])
beacon3 = np.array([16,8])

#Observations
z1 = 10
sigma_z1 = 1
z2 = 5
sigma_z2 = 5
z3 = 9
sigma_z3 = 3


#Combining observations
beacon_all = np.array([beacon1 , beacon2, beacon3])
z_all = np.array([z1,z2,z3])
sigma_z_all = np.array([sigma_z1,sigma_z2,sigma_z3])


#Setup plot
colors = ['red', 'green', 'blue']
axes = plot_setup(num_obs=3)

#Compute & Plot for varying  number of observations
for num_obs in range(1,1+3):
    likelihood = joint_observation_likelihood(z_all[:num_obs],sigma_z_all[:num_obs], beacon_all[:num_obs], gridmap)
    plot_helper(likelihood, colors[:num_obs], z_all[:num_obs],sigma_z_all[:num_obs], beacon_all[:num_obs], num_obs, axes[num_obs-1])
plt.show()
```
